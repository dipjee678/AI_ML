# Visualization with Matplotlib

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
```

## 2D line plot

```python
hours = [2 , 3 , 4, 1 , 5 , 7 ,3] 
days = [1 ,2 , 3 ,4 ,5 ,6 ,7] 

plt.plot(days,hours)
```

```python
plt.xlabel('Days')
plt.ylabel('Hours')
plt.title('Trend of a study time over a week')
plt.plot(days,hours)
```

## Scatter Plot

```python
students = pd.read_csv('student_IQdata.csv')

students
```

```python
#Relation between Study Hours and IQ Score
#positive correlation

plt.xlabel('Study_Hours') 
plt.ylabel('IQ_Score') 
plt.title('Study Hours VS IQ Score') 

plt.scatter(students['Study_Hour'] , students['IQ_Score']) 

```

```python
#Relation between Shoe_Size and IQ Score


plt.xlabel('Shoe_Size') 
plt.ylabel('IQ_Score') 
plt.title('Shoe_Size VS IQ Score') 

plt.scatter(students['Shoe_Size'] , students['IQ_Score']) 

```

```python
#Relation between Chilling_Hoursand IQ Score


plt.xlabel('Chilling_Hours') 
plt.ylabel('IQ_Score') 
plt.title('Chilling_Hours VS IQ Score') 

plt.scatter(students['Chilling_Hours'] , students['IQ_Score'],marker='o',color='red') 
```

## Histogram

## Bar Chart

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# gender wise

gender_group = df.groupby('Gender').size()
gender_group
gender_group.index
gender_group.values
```

```python
plt.bar(gender_group.index,gender_group.values,color=['orange','red'],edgecolor='black')
```

```python
#  Bar chart on real data 

df = pd.read_csv('student_data (1).csv') 

df

```

```python
status = df.groupby('CompletionStatus').size() 

plt.bar(status.index,status.values,color=['green','orange','red'],edgecolor='black')

plt.xlabel('Completion Status') 
plt.ylabel('Count') 
plt.title('Bar Chart of Course completion')
plt.show() 
```

```python
# location

status = df.groupby('Location').size() 

plt.bar(status.index,status.values,edgecolor='black')

plt.xlabel('Location') 
plt.ylabel('Count') 
plt.title('Bar Chart of Location of school')
plt.show() 
```

## Pie chart

```python
status = df.groupby('CompletionStatus').size() 
plt.pie(status,labels=status.index,autopct='%1.1f%%',explode = (0.1,0,0),shadow=True) 
plt.show() 
```
